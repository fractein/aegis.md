# Aegis.md — Explorarea și evaluarea datelor

Acest notebook analizează distribuția datelor de antrenament și rezultatele evaluării modelului.

```python
import csv
from collections import Counter

rows = []
with open('../data/test_dataset.csv', encoding='utf-8') as f:
    rows = list(csv.DictReader(f))

labels = Counter(r['label'] for r in rows)
langs  = Counter(r['language'] for r in rows)

print(f'Total SMS: {len(rows)}')
print(f'Labels:    {dict(labels)}')
print(f'Languages: {dict(langs)}')
```

```python
# Distribuția pe categorii
scam_ro  = sum(1 for r in rows if r['label']=='scam'  and r['language']=='ro')
scam_ru  = sum(1 for r in rows if r['label']=='scam'  and r['language']=='ru')
safe_ro  = sum(1 for r in rows if r['label']=='safe'  and r['language']=='ro')
safe_ru  = sum(1 for r in rows if r['label']=='safe'  and r['language']=='ru')

print(f'Scam RO: {scam_ro} | Scam RU: {scam_ru}')
print(f'Safe RO: {safe_ro} | Safe RU: {safe_ru}')
```
